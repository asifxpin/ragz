```python
from langchain.chat_models import init_chat_model
from langchain.messages import  HumanMessage, SystemMessage, AIMessage
from deep_agent import create_deep_agent
from langchain.tools import tool
from rags import retrieve
from dotenv import load_dotenv
load_dotenv()
```

```python
model = init_chat_model('openai:gpt-4o')
```

```python
agent = create_deep_agent(
    model= model,
    tools=[retrieve],
    system_prompt="you are a friendly assistant, use retrieve for knowledge-base questions.",

)
```

```python
result = agent.invoke({'messages':HumanMessage(content='ماذا ذهب الجميع ليأكلوا من BrightWave؟')})
print(result['messages'][-1].content)
```
